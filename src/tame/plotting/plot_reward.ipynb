```python
from pathlib import Path
from tame.plotting.reward_plotter import Plotter
from tame.utils.utils import ROOT
```

```python

env_name = 'mpe'
# env_name = 'balance'

exp_path = ROOT / 'exps' / env_name

# Select the experiments you want to plot
# The format is:
# {experiment_name: {recalc: bool, agents: int, level: str, name: str}}
# recalc: whether to recalculate the rewards
# agents: number of agents
# level: level of the hierarchy for which to plot the rewards
# name: name to display in the plot

experiments = {
    'ppo_2lev': {'recalc': False, 'agents': 4, 'level': 'bottom_level', 'name': '2PPO'},
    'mpe_oracle': {'recalc': False, 'agents': 4}
}

seeds = 'all'
trace_type = 'mean' # or total
```

```python
plotter = Plotter(exp_path=exp_path, trace_type=trace_type)
```

```python
final_plot = plotter.plot(experiments=experiments, seeds=seeds, smoothing_window=101, plot_ci=False, verbose=0, plot_test=False, scale_by_agents=False, max_x=20000, style='tableau-colorblind10')
```

```python
final_plot.savefig(f'./{env_name}_rewards.svg', format='svg', bbox_inches='tight', dpi=300)
```
